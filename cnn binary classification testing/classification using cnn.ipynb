```python
import cv2
import os

# Paths to folders
dog_folder = "Deep Learning\data\Cat"
cat_folder =C:\Users\USER\Desktop\project1\Deep Learning\data\Dog"

# Read all dog images
dog_images = []
for filename in os.listdir(dog_folder):
    img_path = os.path.join(dog_folder, filename)
    img = cv2.imread(img_path)         # Read the image
    if img is not None:                # Check if image loaded
        img = cv2.resize(img, (128,128))  # Resize to same size
        dog_images.append(img)

# Read all cat images
cat_images = []
for filename in os.listdir(cat_folder):
    img_path = os.path.join(cat_folder, filename)
    img = cv2.imread(img_path)
    if img is not None:
        img = cv2.resize(img, (128,128))
        cat_images.append(img)

print(f"Number of dog images: {len(dog_images)}")
print(f"Number of cat images: {len(cat_images)}")
```
