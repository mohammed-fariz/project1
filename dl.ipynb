```python
import numpy as np
kernel_edge= np.array([
    [-1, -1, -1],
    [-1,  8, -1],
    [-1, -1, -1]
])
# edge detection kernel 
```

```python
kernel_sharp= np.array([
    [0, -1, 0],
    [-1, 5, -1],
    [0, -1, 0]
])
#sharpen kernel
kernel_blur= np.ones((3,3))/9
#blur kernel 
```

```python
output = cv2.filter2D(img, -1, kernel_edge)
```

```python
import matplotlib.pyplot as plt
plt.subplot(1,2,2)
plt.title("Edge Detected Image")
plt.imshow(output, cmap='gray')
plt.axis("off")
```

```python

plt.subplot(1,2,1)
plt.title("Original Grayscale")
plt.imshow(img)
plt.axis("off")
```

```python
b, g, r = cv2.split(img)

edges_r = cv2.Canny(r, 100, 200)
edges_g = cv2.Canny(g, 100, 200)
edges_b = cv2.Canny(b, 100, 200)
```

```python
edges = edges_r + edges_g + edges_b
```

```python

plt.subplot(1,2,2)
plt.title("Edge Detected Image")
plt.imshow(edges)
plt.axis("off")
```
