```python
import os
import shutil
import random

# Paths
source_images = "train/images"
source_labels = "train/labels"
output_dir = "dataset_split"

# Split ratio
train_ratio = 0.7
val_ratio = 0.2
test_ratio = 0.1

# Create folders
for split in ["train", "valid", "test"]:
    os.makedirs(f"{output_dir}/{split}/images", exist_ok=True)
    os.makedirs(f"{output_dir}/{split}/labels", exist_ok=True)

# Get all images
images = [f for f in os.listdir(source_images) if f.endswith(".jpg")]

# Shuffle images
random.shuffle(images)

# Calculate split sizes
total = len(images)
train_end = int(train_ratio * total)
val_end = int((train_ratio + val_ratio) * total)

train_imgs = images[:train_end]
val_imgs = images[train_end:val_end]
test_imgs = images[val_end:]

# Function to copy files
def copy_files(img_list, split):
    for img in img_list:
        label = img.replace(".jpg", ".txt")

        shutil.copy(os.path.join(source_images, img),
                    f"{output_dir}/{split}/images/{img}")

        shutil.copy(os.path.join(source_labels, label),
                    f"{output_dir}/{split}/labels/{label}")

# Perform split
copy_files(train_imgs, "train")
copy_files(val_imgs, "valid")
copy_files(test_imgs, "test")

print("✅ Dataset split completed!")
```
