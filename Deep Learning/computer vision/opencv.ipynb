```python
import cv2
import matplotlib.pyplot as plt
```

```python
img2=cv2.imread("car.jpeg",0)
```

```python
r_image=cv2.resize(img2,(300,300))
plt.imshow(r_image)
```

```python

img3=cv2.imshow('resized_image',r_image)
```

```python
plt.imshow(img2)
```

```python
blur=cv2.GaussianBlur(img,(7,7),0)
plt.imshow(blur)
```

```python
avg_blur=cv2.blur(img,(3,3))
plt.imshow(avg_blur)
```

```python
median_blur=cv2.medianBlur(img,3)

plt.imshow(median_blur)
```

```python
motion_blur=cv2.filter2D(img,-1,(3,3))
plt.imshow(motion_blur)
```

```python
bilaterlfilter_blur=cv2.bilateralFilter(img,d=9,sigmaColor=75,sigmaSpace=75)
plt.imshow(bilaterlfilter_blur)
```

```python
edges=cv2.Canny(img,50,100)
plt.imshow(edges)
```

```python
ret,thre= cv2.threshold(img2, 100, 300, cv2.THRESH_BINARY)
plt.imshow(thre,cmap='gray')
```

```python
THRES=cv2.adaptiveThreshold(img2, 255, cv2.ADAPTIVE_THRESH_MEAN_C,cv2.THRESH_BINARY,11, 2)
plt.imshow(THRES)
```
